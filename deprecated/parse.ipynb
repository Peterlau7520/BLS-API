```python
'''
double for loops; 
second for loop's item B check the first loop one (item A);
first loop for i in range(i+1, len )
second loop  for j in range(0, i+1,  -1)


if less than 4 match -> break

find at least 4 match and if item B and item A's rightest unmatched is a 0, 
item A is item B's parent; record in a dictionary { parent: [child]}


check until reaching 0 (if it's actually the same item, itself is a parent; we can skip.. and it won't exist in our map) 
'''
```

```python
value_pairs = df_ae_merged[['series_id','industry_name']].values.tolist()
```

```python
for i in range( 0,-1,-1):
    print(i)
```

```python
def first_mismatch(s1, s2):
    for i, (c1, c2) in enumerate(zip(s1, s2)):
        if c1 != c2:
            return i
    return min(len(s1), len(s2))  # one is a prefix of the other
```

```python
value_pairs[0][1],value_pairs[0][0]
```

```python
len(value_pairs)
```

```python
hierachy_maps  = {}
for i in range(1, len(value_pairs)):
    for j in range( i-1, -1, -1):
        new_id, child_categroy = value_pairs[i][0],value_pairs[i][1]
        old_id, parent_category = value_pairs[j][0], value_pairs[j][1]
        pos = first_mismatch(old_id, new_id)
        # import pdb
        # pdb.set_trace()
        if pos > 3 and old_id[pos] == '0':
            hierachy_maps[child_categroy]=parent_category
            break

```

```python
'''
next step is do dfs; build the sequences;  

for each key, keep resolving till reaching the parent, and record them in a []

'''
```

```python
hierachy_maps['Financial activities']
```

```python
label_paths = {}
```

```python
for child_key,v in hierachy_maps.items():
    parent_category = v
    path = child_key
    while True: 
        path = parent_category + "&" + path 
        if parent_category in hierachy_maps:
            parent_category = hierachy_maps[parent_category]
        else: 
            break
    # while parent_category in hierachy_maps:
    #     path = parent_category + "&" + path 
    #     parent_category = hierachy_maps[parent_category]
    
    label_paths[child_key] = path
```

```python
df_cat_names = pd.DataFrame(data=[label_paths]).T
```
